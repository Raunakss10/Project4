```python
import os
import pandas as pd
import numpy as np
np.set_printoptions(legacy='1.21')
import requests
import bs4
import lxml
import plotly.express as px
```

```python
recipes = pd.read_csv("RAW_recipes.csv")
interactions = pd.read_csv("interactions.csv")
```

```python
# Replace rating = 0 with NaN
interactions['rating'] = interactions['rating'].replace(0, np.nan)
interactions['rating'].value_counts(dropna=False)
recipes['submitted'] = pd.to_datetime(recipes['submitted'])
interactions['date'] = pd.to_datetime(interactions['date'])
```

```python
df = recipes.merge(
    interactions,
    left_on='id',
    right_on='recipe_id',
    how='left'
)

print("Merged shape:", df.shape)
df.head()
```

```python
avg_rating = (
    df.groupby('id')['rating']
      .mean()
      .rename('average_rating')
)

df = df.merge(avg_rating, on='id', how='left')
```

```python
# Convert string to list
nutrition_cols = [
    'calories', 'total_fat_PDV', 'sugar_PDV',
    'sodium_PDV', 'protein_PDV',
    'saturated_fat_PDV', 'carbs_PDV'
]

df[nutrition_cols] = (
    df['nutrition']
    .str.strip("[]")
    .str.split(",", expand=True)
    .astype(float)
)
df['log_minutes'] = np.log1p(df['minutes'])
```

```python
# basic health indicator based on tags (edit keywords if you want)
healthy_keywords = [
    "healthy", "low-carb", "low carb", "low-fat", "low fat",
    "low calorie", "low-calorie", "high protein", "high-protein",
    "low sugar", "low-sugar"
]

tags_str = df["tags"].astype(str).str.lower()
df["is_healthy_tag"] = tags_str.apply(lambda s: any(k in s for k in healthy_keywords))

# useful transforms
df["log_minutes"] = np.log1p(df["minutes"])
df["log_calories"] = np.log1p(df["calories"])

# choose target column for plots
# if you have average_rating, use it; otherwise use rating
target_col = "average_rating" if "average_rating" in df.columns else "rating"
```

```python
fig1 = px.histogram(
    df.dropna(subset=["calories"]),
    x="log_calories",
    nbins=60,
    title="Distribution of Recipe Calories (log1p transformed)",
    labels={"log_calories": "log(1 + calories)"}
)
fig1
```

```python
fig2 = px.histogram(
    df.dropna(subset=["minutes"]),
    x="log_minutes",
    nbins=60,
    title="Distribution of Recipe Cooking Time (log1p transformed)",
    labels={"log_minutes": "log(1 + minutes)"}
)
fig2
```

```python
healthy = df[df["is_healthy_tag"]]["average_rating"].dropna()
not_healthy = df[~df["is_healthy_tag"]]["average_rating"].dropna()

obs_stat = healthy.mean() - not_healthy.mean()

obs_stat
```

```python
n_perm = 3000
diffs = []

data = df[["average_rating","is_healthy_tag"]].dropna()

for i in range(n_perm):

    shuffled = data.copy()
    shuffled["is_healthy_tag"] = np.random.permutation(shuffled["is_healthy_tag"])

    g1 = shuffled[shuffled["is_healthy_tag"]]["average_rating"]
    g2 = shuffled[~shuffled["is_healthy_tag"]]["average_rating"]

    diff = g1.mean() - g2.mean()

    diffs.append(diff)

diffs = np.array(diffs)
```

```python
p_value = np.mean(diffs <= obs_stat)

print("Observed difference:", obs_stat)
print("p-value:", p_value)
```

```python
import plotly.express as px

fig = px.histogram(
    diffs,
    nbins=50,
    title="Permutation Distribution of Rating Difference"
)

fig.add_vline(x=obs_stat, line_color="red")

fig.show()
```
